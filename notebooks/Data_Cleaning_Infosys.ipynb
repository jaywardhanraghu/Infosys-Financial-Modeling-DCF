```python
#@title Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 200)

plt.style.use("ggplot")
```

```python
#@title Load Excel Workbook

excel_file = "Infosys.xlsx"

xls = pd.ExcelFile(excel_file)

print("Sheets in workbook:")
print(xls.sheet_names)
```

```python
#@title Load Financial Statements

income_statement = pd.read_excel(
    excel_file,
    sheet_name="Profit & Loss"
)

balance_sheet = pd.read_excel(
    excel_file,
    sheet_name="Balance Sheet"
)

cash_flow = pd.read_excel(
    excel_file,
    sheet_name="Cash Flow"
)

print("Income Statement Shape :", income_statement.shape)
print("Balance Sheet Shape    :", balance_sheet.shape)
print("Cash Flow Shape        :", cash_flow.shape)
```

```python
#@title Preview Raw Data

print("Income Statement")
display(income_statement.head())

print("\nBalance Sheet")
display(balance_sheet.head())

print("\nCash Flow")
display(cash_flow.head())
```

```python
#@title Check Missing Values

print("Income Statement")
display(income_statement.isna().sum())

print("\nBalance Sheet")
display(balance_sheet.isna().sum())

print("\nCash Flow")
display(cash_flow.isna().sum())
```

```python
#@title Check Data Types

print("Income Statement")
display(income_statement.dtypes)

print("\nBalance Sheet")
display(balance_sheet.dtypes)

print("\nCash Flow")
display(cash_flow.dtypes)
```

```python
#@title Remove Completely Empty Rows

income_statement = income_statement.dropna(how="all")
balance_sheet = balance_sheet.dropna(how="all")
cash_flow = cash_flow.dropna(how="all")

print("Cleaning Complete")
```

```python
#@title Remove Completely Empty Columns

income_statement = income_statement.dropna(axis=1, how="all")
balance_sheet = balance_sheet.dropna(axis=1, how="all")
cash_flow = cash_flow.dropna(axis=1, how="all")
```

```python
#@title Check Duplicate Rows

print("Income Statement :", income_statement.duplicated().sum())
print("Balance Sheet    :", balance_sheet.duplicated().sum())
print("Cash Flow        :", cash_flow.duplicated().sum())
```

```python
#@title Remove Duplicate Rows

income_statement = income_statement.drop_duplicates()
balance_sheet = balance_sheet.drop_duplicates()
cash_flow = cash_flow.drop_duplicates()
```

```python
#@title Display Dataset Information

print("Income Statement")
income_statement.info()

print("\nBalance Sheet")
balance_sheet.info()

print("\nCash Flow")
cash_flow.info()
```

```python
#@title Export Clean Raw Statements

income_statement.to_csv(
    "income_statement_raw.csv",
    index=False
)

balance_sheet.to_csv(
    "balance_sheet_raw.csv",
    index=False
)

cash_flow.to_csv(
    "cash_flow_raw.csv",
    index=False
)

print("Raw financial statements exported successfully.")
```

```python
#@title Promote First Data Row to Header

def clean_statement(df):

    # Make first row the header
    df.columns = df.iloc[0]

    # Remove the header row
    df = df.iloc[1:].reset_index(drop=True)

    return df

income_statement = clean_statement(income_statement)
balance_sheet = clean_statement(balance_sheet)
cash_flow = clean_statement(cash_flow)
```

```python
#@title Preview Cleaned Statements

print("Income Statement")
display(income_statement.head())

print("\nBalance Sheet")
display(balance_sheet.head())

print("\nCash Flow")
display(cash_flow.head())
```

```python
#@title Rename First Column

income_statement.rename(
    columns={income_statement.columns[0]:"Metric"},
    inplace=True
)

balance_sheet.rename(
    columns={balance_sheet.columns[0]:"Metric"},
    inplace=True
)

cash_flow.rename(
    columns={cash_flow.columns[0]:"Metric"},
    inplace=True
)
```

```python
#@title Keep Historical Years Only

income_statement = income_statement.iloc[:, :11]

balance_sheet = balance_sheet.iloc[:, :11]

cash_flow = cash_flow.iloc[:, :11]
```

```python
#@title Convert Year Columns

for df in [income_statement, balance_sheet, cash_flow]:

    df.columns = [
        col.year if hasattr(col, "year") else col
        for col in df.columns
    ]
```

```python
#@title Convert Financial Values to Numeric

for df in [income_statement, balance_sheet, cash_flow]:

    for col in df.columns[1:]:

        df[col] = (
            df[col]
            .astype(str)
            .str.replace(",", "")
            .str.replace("₹", "")
            .str.strip()
        )

        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
#@title Dataset Information After Cleaning

print(income_statement.info())

print(balance_sheet.info())

print(cash_flow.info())
```

```python
#@title Final Preview

display(income_statement)

display(balance_sheet)

display(cash_flow)
```

```python
#@title Create Metric Lookup Function

def get_metric(df, metric):

    row = df[df["Metric"] == metric]

    if row.empty:
        return None

    return row.iloc[0, 1:]
```

```python
#@title Build Income Statement Dataset

income_df = pd.DataFrame({

    "Year": income_statement.columns[1:],

    "Revenue": get_metric(income_statement, "Sales").values,

    "Operating_Expenses": get_metric(income_statement, "Expenses").values,

    "Operating_Profit": get_metric(income_statement, "Operating Profit").values,

    "Other_Income": get_metric(income_statement, "Other Income").values,

    "Depreciation": get_metric(income_statement, "Depreciation").values,

    "Interest": get_metric(income_statement, "Interest").values,

    "PBT": get_metric(income_statement, "Profit before tax").values,

    "Tax": get_metric(income_statement, "Tax").values,

    "Net_Income": get_metric(income_statement, "Net profit").values,

    "EPS": get_metric(income_statement, "EPS").values

})

income_df
```

```python
#@title Build Balance Sheet Dataset

balance_df = pd.DataFrame({

    "Year": balance_sheet.columns[1:],

    "Share_Capital": get_metric(balance_sheet, "Equity Share Capital").values,

    "Reserves": get_metric(balance_sheet, "Reserves").values,

    "Borrowings": get_metric(balance_sheet, "Borrowings").values,

    "Other_Liabilities": get_metric(balance_sheet, "Other Liabilities").values,

    "Total_Assets": get_metric(balance_sheet, "Total").values,

    "Net_Block": get_metric(balance_sheet, "Net Block").values,

    "CWIP": get_metric(balance_sheet, "Capital Work in Progress").values,

    "Investments": get_metric(balance_sheet, "Investments").values,

    "Other_Assets": get_metric(balance_sheet, "Other Assets").values,

    "Working_Capital": get_metric(balance_sheet, "Working Capital").values,

    "Receivables": get_metric(balance_sheet, "Debtors").values

})

balance_df
```

```python
#@title Build Cash Flow Dataset

cashflow_df = pd.DataFrame({

    "Year": cash_flow.columns[1:],

    "Operating_CF": get_metric(cash_flow, "Cash from Operating Activity").values,

    "Investing_CF": get_metric(cash_flow, "Cash from Investing Activity").values,

    "Financing_CF": get_metric(cash_flow, "Cash from Financing Activity").values,

    "Net_CF": get_metric(cash_flow, "Net Cash Flow").values

})

cashflow_df
```

```python
#@title Verify Data Types

print(income_df.dtypes)

print()

print(balance_df.dtypes)

print()

print(cashflow_df.dtypes)
```

```python
#@title Save Clean Financial Tables

income_df.to_csv("income_statement.csv", index=False)

balance_df.to_csv("balance_sheet.csv", index=False)

cashflow_df.to_csv("cash_flow.csv", index=False)

print("Financial Data Warehouse Created Successfully!")
```
