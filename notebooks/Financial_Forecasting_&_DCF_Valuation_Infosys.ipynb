```python
#@title Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.2f}".format)

plt.style.use("ggplot")
```

```python
#@title Load Historical Financial Data

income_df = pd.read_csv("income_statement.csv")
balance_df = pd.read_csv("balance_sheet.csv")
cashflow_df = pd.read_csv("cash_flow.csv")
```

```python
#@title Historical Financial Summary

financials = pd.DataFrame({

    "Year": income_df["Year"],

    "Revenue": income_df["Revenue"],

    "Operating_Profit": income_df["Operating_Profit"],

    "Depreciation": income_df["Depreciation"],

    "Interest": income_df["Interest"],

    "PBT": income_df["PBT"],

    "Tax": income_df["Tax"],

    "Net_Income": income_df["Net_Income"],

    "Operating_CF": cashflow_df["Operating_CF"],

    "Borrowings": balance_df["Borrowings"],

    "Working_Capital": balance_df["Working_Capital"],

    "Total_Assets": balance_df["Total_Assets"],

    "Share_Capital": balance_df["Share_Capital"],

    "Reserves": balance_df["Reserves"]

})

financials
```

```python
#@title Historical Revenue Growth

financials["Revenue_Growth"] = (
    financials["Revenue"]
    .pct_change()
)

financials[["Year","Revenue","Revenue_Growth"]]
```

```python
#@title Historical Margins

financials["Operating_Margin"] = (
    financials["Operating_Profit"]
    / financials["Revenue"]
)

financials["Tax_Rate"] = (
    financials["Tax"]
    / financials["PBT"]
)

financials["Depreciation_Rate"] = (
    financials["Depreciation"]
    / financials["Revenue"]
)

financials
```

```python
#@title Historical Forecast Drivers

drivers = pd.DataFrame({

    "Metric":[

        "Average Revenue Growth",

        "Average Operating Margin",

        "Average Tax Rate",

        "Average Depreciation Rate"

    ],

    "Historical Average":[

        financials["Revenue_Growth"].mean(),

        financials["Operating_Margin"].mean(),

        financials["Tax_Rate"].mean(),

        financials["Depreciation_Rate"].mean()

    ]

})

drivers["Historical Average"] *=100

drivers.round(2)
```

```python
#@title Historical Forecast Drivers

drivers = pd.DataFrame({

    "Metric":[

        "Average Revenue Growth",

        "Average Operating Margin",

        "Average Tax Rate",

        "Average Depreciation Rate"

    ],

    "Historical Average":[

        financials["Revenue_Growth"].mean(),

        financials["Operating_Margin"].mean(),

        financials["Tax_Rate"].mean(),

        financials["Depreciation_Rate"].mean()

    ]

})

drivers["Historical Average"] *=100

drivers.round(2)
```

#Forecast Assumptions

```python
#@title Forecast Assumptions

forecast = pd.DataFrame({

    "Year":[2027,2028,2029,2030,2031],

    "Revenue_Growth":[

        0.08,

        0.08,

        0.07,

        0.07,

        0.06

    ],

    "Operating_Margin":[

        0.24,

        0.24,

        0.245,

        0.245,

        0.25

    ],

    "Tax_Rate":[

        financials["Tax_Rate"].mean()

    ]*5,

    "Depreciation_Rate":[

        financials["Depreciation_Rate"].mean()

    ]*5

})

forecast
```

#Forecast Income Statement

```python
#@title Forecast Revenue

last_revenue = financials["Revenue"].iloc[-1]

revenues = []

current = last_revenue

for growth in forecast["Revenue_Growth"]:

    current *= (1 + growth)

    revenues.append(current)

forecast["Revenue"] = revenues

forecast
```

```python
#@title Forecast Operating Profit (EBITDA)

forecast["EBITDA"] = (

    forecast["Revenue"]

    * forecast["Operating_Margin"]

)

forecast[["Year","Revenue","EBITDA"]]
```

```python
#@title Forecast Depreciation

forecast["Depreciation"] = (

    forecast["Revenue"]

    * forecast["Depreciation_Rate"]

)

forecast[["Year","Depreciation"]]
```

```python
#@title Forecast EBIT

forecast["EBIT"] = (

    forecast["EBITDA"]

    - forecast["Depreciation"]

)

forecast[["Year","EBITDA","Depreciation","EBIT"]]
```

```python
#@title Forecast Interest Expense

average_interest_rate = (

    financials["Interest"]

    / financials["Borrowings"]

)

average_interest_rate = average_interest_rate.replace(

    [np.inf,-np.inf],

    np.nan

).mean()

forecast["Interest"] = (

    financials["Borrowings"].iloc[-1]

    * average_interest_rate

)

forecast[["Year","Interest"]]
```

```python
#@title Forecast Profit Before Tax

forecast["PBT"] = (

    forecast["EBIT"]

    - forecast["Interest"]

)

forecast[["Year","PBT"]]
```

```python
#@title Forecast Taxes

forecast["Tax"] = (

    forecast["PBT"]

    * forecast["Tax_Rate"]

)

forecast[["Year","Tax"]]
```

```python
#@title Forecast Net Income

forecast["Net_Income"] = (

    forecast["PBT"]

    - forecast["Tax"]

)

forecast[["Year","Net_Income"]]
```

```python
#@title Forecast Income Statement

forecast_income = forecast[[
    "Year",
    "Revenue",
    "EBITDA",
    "Depreciation",
    "EBIT",
    "Interest",
    "PBT",
    "Tax",
    "Net_Income"
]]

forecast_income.round(2)
```

#Forecast Balance Sheet

```python
#@title Historical Balance Sheet Drivers

financials["Debt_to_Revenue"] = (
    financials["Borrowings"] /
    financials["Revenue"]
)

financials["WC_to_Revenue"] = (
    financials["Working_Capital"] /
    financials["Revenue"]
)

financials["Asset_to_Revenue"] = (
    financials["Total_Assets"] /
    financials["Revenue"]
)

financials[[
    "Year",
    "Debt_to_Revenue",
    "WC_to_Revenue",
    "Asset_to_Revenue"
]]
```

```python
#@title Average Historical Drivers

debt_ratio = financials["Debt_to_Revenue"].mean()

wc_ratio = financials["WC_to_Revenue"].mean()

asset_ratio = financials["Asset_to_Revenue"].mean()

print(f"Debt / Revenue        : {debt_ratio:.3f}")
print(f"Working Capital Ratio : {wc_ratio:.3f}")
print(f"Asset Turnover Driver : {asset_ratio:.3f}")
```

```python
#@title Forecast Borrowings

forecast["Borrowings"] = (
    forecast["Revenue"]
    * debt_ratio
)

forecast[["Year","Borrowings"]]
```

```python
#@title Recalculate Interest Expense

forecast["Interest"] = (
    forecast["Borrowings"]
    * average_interest_rate
)

forecast["PBT"] = (
    forecast["EBIT"]
    - forecast["Interest"]
)

forecast["Tax"] = (
    forecast["PBT"]
    * forecast["Tax_Rate"]
)

forecast["Net_Income"] = (
    forecast["PBT"]
    - forecast["Tax"]
)

forecast[[
    "Year",
    "Borrowings",
    "Interest",
    "Net_Income"
]]
```

```python
#@title Forecast Working Capital

forecast["Working_Capital"] = (
    forecast["Revenue"]
    * wc_ratio
)

forecast[[
    "Year",
    "Working_Capital"
]]
```

```python
#@title Forecast Total Assets

forecast["Total_Assets"] = (
    forecast["Revenue"]
    * asset_ratio
)

forecast[[
    "Year",
    "Total_Assets"
]]
```

```python
#@title Forecast Retained Earnings

last_equity = (
    balance_df["Share_Capital"].iloc[-1]
    +
    balance_df["Reserves"].iloc[-1]
)

equity = []

current_equity = last_equity

for profit in forecast["Net_Income"]:

    current_equity += profit

    equity.append(current_equity)

forecast["Equity"] = equity

forecast[[
    "Year",
    "Equity"
]]
```

```python
#@title Forecast Balance Sheet

forecast_balance = forecast[[
    "Year",
    "Borrowings",
    "Working_Capital",
    "Total_Assets",
    "Equity"
]]

forecast_balance.round(2)
```
